```python
porcentaje_null = round((df_peliculas.isna().sum()/len(df_peliculas)) *100,2)
porcentaje_null
```

3. Corrige de la mejor manera que se te ocurra las dos columnas con menor porcentajes de nulos.

```python
columnas_menor = ['rating', 'duration']
for columna in columnas_menor:
    df_peliculas[columna] = df_peliculas[columna].fillna(df_peliculas[columna].mode())
```
