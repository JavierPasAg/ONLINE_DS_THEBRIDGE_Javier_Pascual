#### #1.2  

Recrea la siguiente figura, para ello construye en apartados separados cada gráfico y luego haz un apartado en el que se junten los cuatro.  

Las figuras son:
1. Las ciudades más pobladas, con los numeros de las poblaciones (tendrás que usar texto y recordar que para centrarlos se necesita poner el argumento `ha` a "center"). (Colores: azul, gris, aceituna, marrón, con trasparencia)  

2. Histograma del área total ocupada por las ciudades. (50 bins, fuente = "Bradley Hand ITC", tamaño 18, color azul acero o "steelblue")
3. El scatter de área de agua de un ciudad con su poblacion.
4. El boxplot de la altura de las ciudades.

```python
'''Figura 1
Las ciudades más pobladas, con los numeros de las poblaciones 
(tendrás que usar texto y recordar que para centrarlos se necesita poner el argumento `ha` a "center").
 (Colores: azul, gris, aceituna, marrón, con trasparencia)'''

top5_mas_pobladas = df_california["population_total"].nlargest(5)
top5_mas_pobladas_nombres = df_california.loc[top5_mas_pobladas.index,"city"].tolist()
top5_mas_pobladas = top5_mas_pobladas.tolist()
fig,ax = plt.subplots()
def grafico_barras(ax):
    colores = ['#7f7fff',
            '#bfbfbf',
            '#ff7f7f',
            '#bfbf7f',
            '#d19494']
    columnas = ax.bar(top5_mas_pobladas_nombres, top5_mas_pobladas, color = colores, alpha = 0.8)
    ax.set_title("Ciudades más pobladas de California")
    ax.set_xlabel("Población (miles de habitantes)")
    ax.set_ylabel("Ciudades")
    ax.set_ylim(0, 5000)

    for columna in columnas:
        columna.set_height(columna.get_height()/1000)
        altura = columna.get_height()
        ax.text(
            columna.get_x() + columna.get_width() / 2,
            altura,
            f"{altura:.3f}",
            ha="center",
            va="bottom"
        )
plt.show(grafico_barras(ax))
```

```python
'''Figura 2
Histograma del área total ocupada por las ciudades. 
(50 bins, fuente = "Bradley Hand ITC", tamaño 18, color azul acero o "steelblue")
'''

fig,ax = plt.subplots()
def grafico_histograma(ax):
    ax.hist(df_california["area_total_km2"],bins= 50, color= "steelblue", alpha= 0.8, label= "Hist Area")
    ax.set_ylabel("Número de ciudades",fontname="Bradley Hand ITC")
    ax.set_xlabel(r"Superficie($\bf{km2}$)",fontname="Bradley Hand ITC")
    ax.legend(loc= "upper right",)
plt.show(grafico_histograma(ax))
```

```python
'''Figura 3
El scatter de área de agua de un ciudad con su poblacion.
'''
from matplotlib.ticker import FuncFormatter

fig, ax = plt.subplots()
def grafico_scatter(ax):
    ax.scatter(df_california["area_water_km2"],df_california["population_total"],s = 10, c = "g", alpha = 0.5)
    ax.set_title("Población por kilometro cuadrado de agua")
    ax.set_ylabel("Población (miles habitantes)")
    ax.set_xlabel(r"Área de agua en la ciudad ($km2$)")
    ax.yaxis.set_major_formatter(
    FuncFormatter(lambda x, _: f"{x/1000:.0f}"))
plt.show(grafico_scatter(ax))
```

```python
'''Figura 4
El boxplot de la altura de las ciudades.
'''

fig, ax = plt.subplots()
def grafico_boxplot(ax):
    ax.boxplot(df_california["elevation_m"], vert = False, patch_artist=True,
                boxprops={
                    "facecolor": "red", 
                    "alpha": 0.7},
                medianprops={
                    "color": "black"
                })
    ax.set_title("Diagrama de caja de la Altitud de las ciudades de California")
    ax.set_xlabel(r"Altitud ($m$)")

plt.show(grafico_boxplot(ax))
```

```python
# Todas
fig, axes = plt.subplots(2,2, figsize = (14,14))
grafico_barras(axes[0,0])
grafico_histograma(axes[0,1])
grafico_scatter(axes[1,0])
grafico_boxplot(axes[1,1])
```
